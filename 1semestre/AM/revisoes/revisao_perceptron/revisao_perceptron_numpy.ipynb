# Numpy
Biblioteca utilizada para cálculos matemáticos

```python
import numpy as np 

# é uma funcionalidade particular dele, é importante sempre usá-las pois desenvolve melhor
# vetor [ renda, historicom idade, garantia ] ja normalizados
pesos = np.array([0.3, 0.8, 0.1, 0.5])
bias = -9.0 # pq tem mais features 
```

```python
# vetor [ renda, historicom idade, garantia ] ja normalizados
cliente_teste = np.array([7.0, 8.0, 4.0, 6.0])

# Cálculo Z (Equação de decisão)
# np.dot: independente da qtd de entradas faz produto escalar entre os dois vetores, ou seja, multiplica cada elemento do vetor cliente_teste pelo elemento correspondente do vetor pesos e depois soma tudo
z = np.dot(cliente_teste, pesos) + bias

# função de ativação
decisao = 1 if z >= 0 else 0 
print(f"O valor de Z é: {z:.2f}")
print(f"A decisão é: {'Aprovado' if decisao == 1 else 'Reprovado'}")
```

# Numpy em lote

```python
clientes_lote = np.array([
    [7.0, 8.0, 4.0, 6.0], # Cliente A - bom perfil
    [3.0, 2.0, 5.0, 0.0], # CLiente B - risco alto, sem garantia 
    [5.0, 5.0, 5.0, 10.0] # Cliente C - perfil médio
])

z_lote = np.dot(clientes_lote, pesos) + bias 

decisoes_lote = (z_lote >= 0).astype(int) # vetor de decisões para cada cliente no lote

for i, decisao in enumerate(decisoes_lote):
    print(f"Cliente {i+1}: Z = {z_lote[i]:.2f} - {'Aprovado' if decisao == 1 else 'Reprovado'}")
```

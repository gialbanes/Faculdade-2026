# Entradas

```python
# 0 a 10 - representando faixas salariais 
renda_mensal = 2.0 

# 0 a 10 - bom pagador
historico_pagador = 10.0
```

# Pesos

```python
peso_renda = 0.3
peso_historico = 0.8

# Bias (a resistência natural em emprestar dinheiro), ou seja, de cara o banco não quer emprestar
# não pode ser muito positivo, pois tenda a falar sim pra maioria dos dados, então não teria resistência 
bias = -7.0
```

# Cálculo Z - equação de decisão

```python
z = (renda_mensal * peso_renda) + (historico_pagador * peso_historico) + bias 

print(f"O valor interno gerado pelo Perceptron = {z:.2f}")
```

# Função de ativação (Degrau ou Step)

```python
if z >= 0:
    decisao = 1
    motivo = "Aprovado, bom pagador!"
else:
    decisao = 0
    motivo = "Reprovado, risco muito alto!"

print(f"Decisão final: {decisao}")    
print(f"Motivo: {motivo}")    
```

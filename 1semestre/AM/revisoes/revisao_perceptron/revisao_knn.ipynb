# KNN - Manual

```python
# Raiz quadrada para calcular a distancia
import math

import pandas as pd

# Classificação com KNN
from sklearn.neighbors import KNeighborsClassifier
```

```python
# Euclidiana
# Cliente A e Cliente B : renda e historico de pagament
p1 = (2.0, 3.0) # Clite A renda baixa, historico ruim
p2 = (8.0, 7.0) # CLiente B renda alta, historico bom

# Diferença em cada dimensão (cateto do triangulo retangulo)
dx = p1[0] - p2[0]
dy = p1[1] - p2[1]

# Distancia = Hipotenusa (Pitagoras)
distancia = math.sqrt(dx**2 + dy**2)
distancia_manhattan = abs(dx) + abs(dy)

print(f"Euclidiana: {distancia:.2f}")
print(f"Manhattan: {distancia_manhattan:.2f}")
print()
print("Euclidiana: distancia em linha reta.")
print("Manhattan: soma dos passos em grade (horizontal + vertical).")
```

```python
clientes = [
    {"renda": 8.0, "historico": 9.0, "aprovado": 1},
    {"renda": 3.0, "historico": 3.0, "aprovado": 0},
    {"renda": 6.0, "historico": 7.0, "aprovado": 1},
    {"renda": 2.0, "historico": 5.0, "aprovado": 0},
    {"renda": 9.0, "historico": 8.0, "aprovado": 1},
]
```

```python
# novo cliente
novo = (5.0, 6.0)

for c in clientes:
    dx = novo[0] - c['renda'] # x - entradas
    dy = novo[1] - c['historico'] # y - entradas
    dist = math.sqrt(dx**2 + dy**2)
    print(f" dist={dist:.2f} | renda={c['renda']}, hist={c['historico']}, aprovado={c['aprovado']}")
```

```python
# KNN - votação
k = 3 # 3 ou 5 é votação da maioria 

#calcula a distância e guarda junto com o resultado 
clientes_com_dist = []
for c in clientes:
    dx = novo[0] - c['renda']
    dy = novo[1] - c['historico']
    dist = math.sqrt(dx**2 + dy**2)
    clientes_com_dist.append({"dist": dist, **c})

# ordena do mais próximo para o mais distante para fazer a votação
clientes_com_dist.sort(key=lambda x:x["dist"]) # pega um por um da distancia e organizar o mais próximo para o mais distante

# pega os k primeiros (os mais próximos)
vizinhos = clientes_com_dist[:k]
votos = [v["aprovado"] for v in vizinhos] 

print(f"Novo cliente: renda={novo[0]}, historico={novo[1]}")
print(f"\nOs {k} vizinhos mais próximos:")
for v in vizinhos:
    label = "Aprovado" if v["aprovado"] == 1 else "Reprovado"
    print(f"  dist={v['dist']:.2f} = {label}")

# votação - maior decide 
# não importa se o vizinho mais proximo mudou o valor, o que decide é a maioria dos vizinhos
resultado = 1 if sum(votos) > k // 2 else 0
print(f"\nVotos: {votos}")
print(f"Previsão: {'Aprovado' if resultado == 1 else 'Reprovado'}")
```

# Fazer o KNN com scikit learn

```python
df = pd.DataFrame([
    {"renda": 8.0, "historico": 9.0, "aprovado": 1},
    {"renda": 3.0, "historico": 3.0, "aprovado": 1},
    {"renda": 8.0, "historico": 9.0, "aprovado": 1},
    {"renda": 8.0, "historico": 9.0, "aprovado": 1},
    {"renda": 8.0, "historico": 9.0, "aprovado": 1},
    {"renda": 8.0, "historico": 9.0, "aprovado": 1},
    {"renda": 8.0, "historico": 9.0, "aprovado": 1},
])

# separa as features (X) e o rótulo (Y)
X = df[["renda", "historico"]]
y = df["aprovado"]

#eucliana ou manhattan pro scikit-learn entender qual vamos usar
metrica = "euclidean"

modelo = KNeighborsClassifier(n_neighbors=3, metric=metrica)
modelo.fit(X, y)

# novo cliente
novo_cliente = pd.DataFrame([{"renda": 5.0, "historico": 6.0}])
previsao = modelo.predict(novo_cliente)

print(f"Métrica usada: {metrica}")
print(f"Previsão: {'Aprovado' if previsao[0] == 1 else 'Reprovado'}")
```
